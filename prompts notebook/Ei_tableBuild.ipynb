```python
import numpy as np
import pandas as pd
```

```python
df_GEI = pd.read_csv(r"C:\Users\ttjrb\OneDrive\Desktop\University Important\PROJECTS\NBA data analyzed\database\GEI_scraped.csv")
df_metrics = pd.read_csv(r"/database/Ei_metrics_beforeGEI.csv")
```

```python
def match_gei(row):
    matched = df_GEI[
        (df_GEI["Date"] == row["DATE"]) &
        (df_GEI["Score"].str.contains(row["TEAM1"], case=False, na=False)) &
        (df_GEI["Score"].str.contains(row["TEAM2"], case=False, na=False))
    ]
    return matched["Excitment"].mean() if not matched.empty else 0
```

```python
df_metrics["GEI"] = df_metrics.apply(match_gei, axis=1)
```
